**Exploratory data analysis**

```python
import cleaning_Q7_Q8 as clean
import matplotlib.pyplot as plt
import seaborn as sns

q7_word_df, q8_by_label, q7_by_label, q7_counts, df = clean.clean_data("cleaned_data_combined_modified.csv")

```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
#response distribution for Q7
sns.barplot(x=q7_counts.head(10).index, y=q7_counts.head(10).values, ax=axes[0])
axes[0].set_title("Top 10 'Who does this food remind you of?' Responses")
axes[0].set_ylabel("Count")
axes[0].set_xlabel("Response")
axes[0].tick_params(axis="x", rotation=90)

#response distribution for Q8
sns.barplot(
    x=df["Q8: How much hot sauce would you add to this food item?"].value_counts().index,
    y=df["Q8: How much hot sauce would you add to this food item?"].value_counts().values,
    ax=axes[1]
)
axes[1].set_title("Hot Sauce Preferences Distribution")
axes[1].set_ylabel("Count")
axes[1].set_xlabel("Preference")
axes[1].tick_params(axis="x", rotation=90)
```

```python
#q7 heatmap by target value
plt.figure(figsize=(12, 6))
sns.heatmap(q7_word_df, cmap="Greens", annot=True, fmt="g")
plt.title("Q7 Word Counts by Label")
plt.xlabel("Label")
plt.ylabel("Word")
plt.show()
```

```python
#q8 heatmap by target value
plt.figure(figsize=(12, 6))
sns.heatmap(q8_by_label, cmap="Oranges", annot=True, fmt="g")
plt.title("Q8 Responses by Label")
plt.xlabel("Preference")
plt.ylabel("Label")
plt.show()
```

From Q7, some patterns that were noticed are that pizza is strongly associated with friends and strangers, moreso than the other 2. It's also notable that shawarma is associated to strangers more than the other 2 and sushi is more associated to parents than the other 2.

From Q8, some patterns that were noticed is that sushi has many more responses with no hot sauce, shawarma has many more responses with medium amounts of hot sauce, and pizza does not really have anything specific that stands out.
